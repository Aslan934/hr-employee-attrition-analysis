## 2. Attrition Analysis

### Task 3. Ümumi Attrition göstəricisi

- İşdən ayrılan və şirkətdə qalan əməkdaşların sayını tapın.
- Ümumi Attrition rate-i hesablayın və vizuallaşdırın.

```python
# İşdən ayrılan və şirkətdə qalan əməkdaşların sayı
attrition_counts = df["Attrition"].value_counts()

print("İşdən ayrılan əməkdaşların sayı:", attrition_counts.get("Yes", 0))
print("Şirkətdə qalan əməkdaşların sayı:", attrition_counts.get("No", 0))


# Ümumi Attrition Rate
attrition_rate = (df["Attrition"].eq("Yes").sum() / len(df)) * 100

print(f"\nÜmumi Attrition Rate: {attrition_rate:.2f}%")

plt.figure(figsize=(6, 6))

plt.pie(
    attrition_counts,
    labels=attrition_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Overall Attrition Rate")
plt.show()
```

### Task 4. Department üzrə Attrition

- Hər Department üzrə əməkdaş sayını və Attrition rate-i hesablayın.
- Ən yüksək Attrition rate olan Department-i müəyyən edin və nəticəni vizuallaşdırın.

```python
# Department üzrə ümumi əməkdaş sayı
employee_count = df["Department"].value_counts()

# hər department üzrə əməkdaş sayı
df['Department'].value_counts()

# İşdən ayrılan əməkdaşları seçirik
attrition_yes = df[df["Attrition"] == "Yes"]

# Department üzrə ayrılan əməkdaşların sayı
attrition_count = attrition_yes["Department"].value_counts()

# Attrition Rate
department_analysis = pd.DataFrame({
    "Employee_Count": employee_count,
    "Attrition_Count": attrition_count
})

# Nəticələri birləşdiririk
department_analysis = pd.DataFrame({
    "Employee_Count": employee_count,
    "Attrition_Count": attrition_count
})

department_analysis["Attrition_Count"] = department_analysis["Attrition_Count"].fillna(0)

# Attrition Rate
department_analysis["Attrition_Rate"] = (
    department_analysis["Attrition_Count"] /
    department_analysis["Employee_Count"] * 100
).round(2)

display(department_analysis.reset_index())
```

```python
highest_attrition = department_analysis.loc[
    department_analysis["Attrition_Rate"].idxmax()
]

print("Ən yüksək Attrition Rate olan Department:")
highest_attrition
department_analysis["Attrition_Rate"].plot(kind="bar", figsize=(8,5))

plt.title("Attrition Rate by Department")
plt.xlabel("Department")
plt.ylabel("Attrition Rate (%)")
plt.show()
```

### Task 5. Job Role üzrə Attrition

- Hər Job Role üzrə Attrition rate-i hesablayın.
- Attrition rate-i ən yüksək olan Job Role-ları vizuallaşdırın.

```python
# Hər Job Role üzrə ümumi əməkdaş sayı
job_role_count = df.groupby("Job_Role")["Employee_ID"].count()

# İşdən ayrılan əməkdaşlar
attrition_yes = df[df["Attrition"] == "Yes"]

# Hər Job Role üzrə ayrılan əməkdaş sayı
job_role_attrition = attrition_yes.groupby("Job_Role")["Employee_ID"].count()

# Nəticələri birləşdiririk
job_role_analysis = pd.DataFrame({
    "Employee_Count": job_role_count,
    "Attrition_Count": job_role_attrition
})

# Missing values varsa 0 edirik
job_role_analysis["Attrition_Count"] = job_role_analysis["Attrition_Count"].fillna(0)

# Attrition Rate hesablayırıq
job_role_analysis["Attrition_Rate"] = (
    job_role_analysis["Attrition_Count"] /
    job_role_analysis["Employee_Count"] * 100
).round(2)

display(job_role_analysis.sort_values("Attrition_Rate", ascending=False))
```

```python
top_roles = job_role_analysis.sort_values(
    "Attrition_Rate", ascending=False
).head(5)

plt.figure(figsize=(10, 5))

sns.barplot(
    x=top_roles.index,
    y=top_roles["Attrition_Rate"]
)

plt.title("Top 5 Job Roles with Highest Attrition Rate")
plt.xlabel("Job Role")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=45)

plt.show()
```

### Task 6. Overtime və Attrition

- Overtime edən və etməyən əməkdaşların Attrition rate-lərini müqayisə edin.
- Nəticəni uyğun qrafiklə göstərin.

```python
# Ümumi əməkdaş sayı
overtime_total = df.groupby("Overtime")["Employee_ID"].count()

# İşdən ayrılan əməkdaşların sayı
overtime_yes = df[df["Attrition"] == "Yes"].groupby("Overtime")["Employee_ID"].count()

# Attrition Rate
overtime_attrition_rate = (
    overtime_yes / overtime_total * 100
).round(2)

print("Overtime üzrə Attrition Rate:")
print(overtime_attrition_rate)
```

```python
plt.figure(figsize=(7, 5))

overtime_attrition_rate.plot(kind="bar")

plt.title("Attrition Rate by Overtime")
plt.xlabel("Overtime")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=0)

plt.show()
```

### Task 8. Age Groups

- Əməkdaşları uyğun Age Group-lara bölün.
- Hər Age Group üzrə Attrition rate-i hesablayın və vizuallaşdırın.

```python
# Ilk once emekdaslarin yashlarini unique olaraq kontrol edirik
np.sort(df['Age'].unique())

# Uygun yash qruplari
labels = ["18-25","26-35","36-45","45-55","56+"]
bins = [0,25,35,45,55,100]


# Bolmek ucun cut funksiyasindan istifade edirik
df["Age_Group"]=pd.cut(
    df['Age'],
    bins=bins,
    labels=labels
)

display(df[["Age", "Age_Group"]].head())
```

```python
#Yash qruplarinin sayini tapiriq
age_group_counts = df['Age_Group'].value_counts()

# Yash qruplarina gore ishden ayrilan emekdash sayi tapiriq
age_group_attrition = df[df['Attrition']=='Yes'].groupby('Age_Group')['Employee_ID'].count()

# Yash qrupuna gore attrition rate tapiriq
age_group_att_rate = (age_group_attrition/age_group_counts*100).round(2)

print("Attrition Rate by age groups:")
display(age_group_att_rate)
```

```python
#Visuallasdirma
plt.figure(figsize=(8, 5))

age_group_att_rate.plot(kind="bar")

plt.title("Attrition Rate by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=0)

plt.show()
```

## 4. Relationships və Yekun Nəticələr

### Task 13. Chi-Square Test

- Department və Attrition arasında əlaqə olub-olmadığını Chi-Square Test ilə yoxlayın.
- Alınan nəticəni şərh edin.

```python
from scipy.stats import chi2_contingency

# Contingency table
table = pd.crosstab(df["Department"], df["Attrition"])

print("Contingency Table:")
display(table)

# Chi-Square Test
chi2, p_value, dof, expected = chi2_contingency(table)

print("Chi-Square:", chi2)
print("p-value:", p_value)
print("Degrees of Freedom:", dof)

if p_value < 0.05:
    print("Nəticə: Department və Attrition arasında statistik olaraq əhəmiyyətli əlaqə var.")
else:
    print("Nəticə: Department və Attrition arasında statistik olaraq əhəmiyyətli əlaqə yoxdur.")
```

Nəticə: Department və Attrition arasında statistik olaraq əhəmiyyətli əlaqə yoxdur.

### Task 14. Correlation Analysis

- Analiz üçün uyğun dəyişənlərdən correlation matrix yaradın.
- Correlation heatmap hazırlayın və Attrition ilə ən güclü əlaqəsi olan göstəriciləri müəyyən edin.

```python
# Attrition-i numeric formata çeviririk
df["Attrition_Num"] = df["Attrition"].map({
    "No": 0,
    "Yes": 1
})

# Numeric dəyişənləri seçirik
numeric_df = df.select_dtypes(include="number")

# Correlation matrix
correlation_matrix = numeric_df.corr()

display(correlation_matrix)
```

```python
plt.figure(figsize=(14, 10))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Matrix Heatmap")
plt.show()
```
